```python

import networkx as nx
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python

# Load the Karate Club social network
G = nx.karate_club_graph()

# Display network information
print("Dataset: Karate Club Social Network")
print("Number of Nodes:", G.number_of_nodes())
print("Number of Edges:", G.number_of_edges())
print("Is the graph connected?", nx.is_connected(G))
```

```python

plt.figure(figsize=(10, 7))

pos = nx.spring_layout(G, seed=42)

nx.draw_networkx(
    G,
    pos,
    node_color="skyblue",
    node_size=700,
    edge_color="gray",
    with_labels=True,
    font_size=9
)

plt.title("Karate Club Social Network")
plt.axis("off")
plt.show()
```

```python

# Calculate centrality measures

degree = nx.degree_centrality(G)

betweenness = nx.betweenness_centrality(G)

closeness = nx.closeness_centrality(G)

pagerank = nx.pagerank(G)

# Create a DataFrame
centrality_df = pd.DataFrame({
    "Node": list(G.nodes()),
    "Degree Centrality": [degree[n] for n in G.nodes()],
    "Betweenness Centrality": [betweenness[n] for n in G.nodes()],
    "Closeness Centrality": [closeness[n] for n in G.nodes()],
    "PageRank": [pagerank[n] for n in G.nodes()]
})

# Display the first 10 rows
centrality_df.head(10)
```

```python

# Find top 5 nodes using each centrality measure

measures = [
    "Degree Centrality",
    "Betweenness Centrality",
    "Closeness Centrality",
    "PageRank"
]

for measure in measures:

    print("\nTop 5 nodes based on", measure)

    top_nodes = centrality_df.sort_values(
        by=measure,
        ascending=False
    ).head(5)

    print(top_nodes[["Node", measure]])
```

```python

# Plot top 5 nodes for each centrality measure

for measure in measures:

    top_nodes = centrality_df.sort_values(
        by=measure,
        ascending=False
    ).head(5)

    plt.figure(figsize=(8, 4))

    sns.barplot(
        data=top_nodes,
        x="Node",
        y=measure
    )

    plt.title("Top 5 Nodes - " + measure)
    plt.show()
```

```python

# Rank nodes based on all centrality measures

for measure in measures:
    centrality_df[measure + " Rank"] = (
        centrality_df[measure]
        .rank(ascending=False, method="min")
        .astype(int)
    )

# Display top 10 nodes based on degree centrality
ranking_table = centrality_df.sort_values(
    by="Degree Centrality",
    ascending=False
)

display(ranking_table.head(10).round(4))
```

```python

# Identify top 5 nodes based on PageRank

top_5 = centrality_df.nlargest(5, "PageRank")["Node"].tolist()

# Assign different colors to influential nodes
node_colors = [
    "red" if node in top_5 else "skyblue"
    for node in G.nodes()
]

plt.figure(figsize=(11, 8))

pos = nx.spring_layout(G, seed=42)

nx.draw_networkx(
    G,
    pos,
    node_color=node_colors,
    node_size=750,
    edge_color="gray",
    with_labels=True,
    font_size=9
)

plt.title("Top 5 Influential Nodes Using PageRank")
plt.axis("off")
plt.show()

print("Top 5 influential nodes:", top_5)
```

```python

# Compare top 5 nodes from all centrality measures

for measure in measures:
    print("\nTop 5 nodes based on", measure)

    result = centrality_df.nlargest(5, measure)

    print(
        result[["Node", measure]]
        .to_string(index=False)
    )
```

```python

# Compare top 5 node overlap

top_nodes_dict = {}

for measure in measures:
    top_nodes_dict[measure] = set(
        centrality_df.nlargest(5, measure)["Node"]
    )

print("Top 5 Node Overlap Analysis\n")

for i in range(len(measures)):
    for j in range(i + 1, len(measures)):

        m1 = measures[i]
        m2 = measures[j]

        common = (
            top_nodes_dict[m1] &
            top_nodes_dict[m2]
        )

        print(m1, "vs", m2)
        print("Common Nodes:", sorted(common))
        print("Number of Common Nodes:", len(common))
        print("--------------------------------")
```

```python

# Calculate overall rank using all four measures

rank_columns = [
    measure + " Rank" for measure in measures
]

centrality_df["Average Rank"] = (
    centrality_df[rank_columns].mean(axis=1)
)

# Lower average rank means more consistently influential
overall_ranking = centrality_df.sort_values(
    by="Average Rank"
)

print("Overall Top 10 Influential Nodes")

display(
    overall_ranking[
        ["Node"] + rank_columns + ["Average Rank"]
    ].head(10).round(3)
)
```

```python

# Correlation between centrality measures

correlation = centrality_df[measures].corr(method="spearman")

print("Spearman Rank Correlation:")
display(correlation.round(3))

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Correlation Between Centrality Measures")
plt.show()
```
